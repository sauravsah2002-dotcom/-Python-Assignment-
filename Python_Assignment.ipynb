Question 1: Explain the difference between:
● "r", "w", "a", "rb"
● Why context managers (with open()) are preferred
● What happens if a file is not properly closed in a production system

Answer:
a.
"r"(read) is used to open files for reading.
e.g: open("data.text","r");
"w"(write) is used to open the file for writing.
e.g: open("data.text","w");
"a"(append) is used to open the file for writing, but append new content at the end instead of overwritting.
e.g: open("data.text","a");
"rb"(read binary) is used to open the file in binary mode foe reading.
e.g: open("image.png","rb").

b.Context manager are favariable due to following reasons:
1.Automatic closing: The file is closed as soon as the block ends, even if an error occurs.

2.Cleaner code: No need to explicitly call file.close().

3.Safety: Prevents resource leaks and ensures proper release of system resources.

c. If A file is not closed properly in a production system, then following problems ocurs--
Resource leaks: The file handle remains open, consuming system resources.

Data corruption: Buffers may not be flushed, so written data might not be saved correctly.

File locks: On some systems, the file may remain locked, preventing other processes from accessing it.

4. Explain impact of dirty data on ML performance ?

Answer : Key Impacts of Dirty Data on ML Performance ---

a. Reduced Accuracy
Example: Mislabeling “fraudulent” transactions as “legitimate” reduces fraud detection accuracy.

b. Bias and Skewed Predictions
Example: If customer data is missing for certain regions, predictions will favor regions with more complete records.

c. Poor Generalization
Example: Duplicate rows inflate certain patterns, making the model overfit.

d. Increased Computational Cost
Cleaning dirty data late in the pipeline wastes resources.

e. Unreliable Decision-Making
In critical domains (e.g; finance, healthcare), dirty data can lead to dangerous mispredictions.
